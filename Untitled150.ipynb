```python
import os
import json
import yaml
import joblib
import warnings
import numpy as np
import pandas as pd
import tensorflow as tf

from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

warnings.filterwarnings("ignore")

# ============================================================
# 1. PROJECT PATH CONFIGURATION
# ============================================================

PROJECT_DIR = r"C:\Users\sagni\Downloads\AI-Based Government Fund Allocation and Utilization Analysis and Forecasting"

CSV_PATH = os.path.join(
    PROJECT_DIR,
    "RS_Session_270_AU_2397_B.ii_.csv"
)

os.makedirs(PROJECT_DIR, exist_ok=True)

H5_PATH = os.path.join(PROJECT_DIR, "government_fund_model.h5")
PKL_PATH = os.path.join(PROJECT_DIR, "government_fund_model.pkl")
YAML_PATH = os.path.join(PROJECT_DIR, "model_config.yaml")
JSON_PATH = os.path.join(PROJECT_DIR, "model_results.json")

# ============================================================
# 2. LOAD DATASET
# ============================================================

print("\nLoading dataset...")

df = pd.read_csv(CSV_PATH)

print("\nOriginal Dataset:")
print(df.head())

print("\nDataset Information:")
print(df.info())

print("\nMissing Values:")
print(df.isnull().sum())

# ============================================================
# 3. IDENTIFY COLUMNS
# ============================================================

print("\nAvailable Columns:")
print(df.columns.tolist())

# Standardize column names
df.columns = df.columns.str.strip()

# Identify year column
year_col = next(
    (col for col in df.columns if "year" in col.lower()),
    df.columns[0]
)

# Identify allocated funds column
allocated_col = next(
    (col for col in df.columns if "allocat" in col.lower()),
    None
)

# Identify utilized funds column
utilized_col = next(
    (
        col for col in df.columns
        if "utilis" in col.lower() or "utiliz" in col.lower()
    ),
    None
)

if allocated_col is None or utilized_col is None:
    raise ValueError(
        "Could not identify fund allocation/utilization columns. "
        "Please check the CSV column names."
    )

print("\nDetected Columns:")
print("Year:", year_col)
print("Allocated Funds:", allocated_col)
print("Utilized Funds:", utilized_col)

# ============================================================
# 4. DATA PREPROCESSING
# ============================================================

print("\nPreprocessing dataset...")

# Extract starting year from financial year
df["Year_Numeric"] = (
    df[year_col]
    .astype(str)
    .str.extract(r"(\d{4})")[0]
)

df["Year_Numeric"] = pd.to_numeric(
    df["Year_Numeric"],
    errors="coerce"
)

# Convert financial values to numeric
for col in [allocated_col, utilized_col]:

    df[col] = pd.to_numeric(
        df[col].astype(str).str.replace(",", "", regex=False),
        errors="coerce"
    )

# Keep complete historical records for training
train_df = df.dropna(
    subset=[
        "Year_Numeric",
        allocated_col,
        utilized_col
    ]
).copy()

train_df = train_df.sort_values("Year_Numeric")

if len(train_df) < 2:
    raise ValueError(
        "At least two complete historical records are required."
    )

print("\nCleaned Dataset:")
print(train_df)

# ============================================================
# 5. FEATURE AND TARGET SELECTION
# ============================================================

X = train_df[["Year_Numeric"]].values

y = train_df[
    [allocated_col, utilized_col]
].values

# ============================================================
# 6. FEATURE SCALING
# ============================================================

x_scaler = StandardScaler()
y_scaler = StandardScaler()

X_scaled = x_scaler.fit_transform(X)
y_scaled = y_scaler.fit_transform(y)

# Save preprocessing objects
joblib.dump(
    {
        "x_scaler": x_scaler,
        "y_scaler": y_scaler,
        "feature_columns": ["Year_Numeric"],
        "target_columns": [allocated_col, utilized_col]
    },
    os.path.join(PROJECT_DIR, "scalers.pkl")
)

# ============================================================
# 7. RANDOM FOREST MODEL
# ============================================================

print("\nTraining Random Forest Regressor...")

rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

rf_model.fit(X, y)

# Save Random Forest model
joblib.dump(rf_model, PKL_PATH)

print("Random Forest model saved successfully!")

# ============================================================
# 8. RANDOM FOREST PREDICTIONS
# ============================================================

rf_predictions = rf_model.predict(X)

rf_mae = mean_absolute_error(y, rf_predictions)

rf_mse = mean_squared_error(y, rf_predictions)

rf_rmse = np.sqrt(rf_mse)

rf_r2 = r2_score(y, rf_predictions)

print("\nRandom Forest Training Metrics:")
print("MAE:", rf_mae)
print("MSE:", rf_mse)
print("RMSE:", rf_rmse)
print("R2 Score:", rf_r2)

# ============================================================
# 9. NEURAL NETWORK MODEL
# ============================================================

print("\nTraining Neural Network...")

tf.random.set_seed(42)

nn_model = tf.keras.Sequential([

    tf.keras.layers.Input(shape=(1,)),

    tf.keras.layers.Dense(
        32,
        activation="relu"
    ),

    tf.keras.layers.Dense(
        16,
        activation="relu"
    ),

    tf.keras.layers.Dense(
        2,
        activation="linear"
    )

])

nn_model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.01
    ),
    loss="mse",
    metrics=["mae"]
)

history = nn_model.fit(
    X_scaled,
    y_scaled,
    epochs=200,
    batch_size=2,
    verbose=0
)

# Save neural network model
nn_model.save(H5_PATH)

print("Neural Network model saved successfully!")

# ============================================================
# 10. NEURAL NETWORK PREDICTIONS
# ============================================================

nn_predictions_scaled = nn_model.predict(
    X_scaled,
    verbose=0
)

nn_predictions = y_scaler.inverse_transform(
    nn_predictions_scaled
)

nn_mae = mean_absolute_error(
    y,
    nn_predictions
)

nn_mse = mean_squared_error(
    y,
    nn_predictions
)

nn_rmse = np.sqrt(nn_mse)

nn_r2 = r2_score(
    y,
    nn_predictions
)

print("\nNeural Network Training Metrics:")
print("MAE:", nn_mae)
print("MSE:", nn_mse)
print("RMSE:", nn_rmse)
print("R2 Score:", nn_r2)

# ============================================================
# 11. FUTURE FUND FORECASTING
# ============================================================

print("\nGenerating Future Predictions...")

last_year = int(
    train_df["Year_Numeric"].max()
)

future_years = np.arange(
    last_year + 1,
    last_year + 6
).reshape(-1, 1)

# Random Forest predictions
future_rf_predictions = rf_model.predict(
    future_years
)

# Neural Network predictions
future_scaled = x_scaler.transform(
    future_years
)

future_nn_scaled = nn_model.predict(
    future_scaled,
    verbose=0
)

future_nn_predictions = y_scaler.inverse_transform(
    future_nn_scaled
)

future_df = pd.DataFrame({

    "Year": future_years.flatten(),

    "RF_Allocated_Funds":
        future_rf_predictions[:, 0],

    "RF_Utilized_Funds":
        future_rf_predictions[:, 1],

    "NN_Allocated_Funds":
        future_nn_predictions[:, 0],

    "NN_Utilized_Funds":
        future_nn_predictions[:, 1]

})

future_df.to_csv(
    os.path.join(
        PROJECT_DIR,
        "future_predictions.csv"
    ),
    index=False
)

print("\nFuture Predictions:")
print(future_df)

# ============================================================
# 12. SAVE YAML CONFIGURATION
# ============================================================

config = {

    "project_name":
        "AI-Based Government Fund Allocation and Utilization Analysis and Forecasting",

    "dataset":
        os.path.basename(CSV_PATH),

    "dataset_rows":
        int(len(df)),

    "training_rows":
        int(len(train_df)),

    "features":
        ["Year_Numeric"],

    "targets":
        [allocated_col, utilized_col],

    "models": {

        "RandomForest": {
            "n_estimators": 100,
            "random_state": 42
        },

        "NeuralNetwork": {
            "hidden_layers": [32, 16],
            "activation": "relu",
            "optimizer": "adam",
            "epochs": 200,
            "batch_size": 2
        }

    },

    "model_files": {
        "random_forest":
            "government_fund_model.pkl",

        "neural_network":
            "government_fund_model.h5",

        "scalers":
            "scalers.pkl"
    }

}

with open(YAML_PATH, "w", encoding="utf-8") as file:

    yaml.safe_dump(
        config,
        file,
        sort_keys=False
    )

print("\nYAML configuration saved!")

# ============================================================
# 13. SAVE JSON RESULTS
# ============================================================

results = {

    "project":
        "Government Fund Allocation and Utilization Forecasting",

    "dataset_summary": {

        "total_records":
            int(len(df)),

        "complete_records":
            int(len(train_df)),

        "missing_records":
            int(len(df) - len(train_df))

    },

    "RandomForest": {

        "MAE":
            float(rf_mae),

        "MSE":
            float(rf_mse),

        "RMSE":
            float(rf_rmse),

        "R2":
            float(rf_r2)

    },

    "NeuralNetwork": {

        "MAE":
            float(nn_mae),

        "MSE":
            float(nn_mse),

        "RMSE":
            float(nn_rmse),

        "R2":
            float(nn_r2)

    },

    "future_predictions":
        future_df.to_dict(
            orient="records"
        )

}

with open(JSON_PATH, "w", encoding="utf-8") as file:

    json.dump(
        results,
        file,
        indent=4
    )

print("\nJSON results saved!")

# ============================================================
# 14. SAVE TRAINING HISTORY
# ============================================================

history_df = pd.DataFrame(
    history.history
)

history_df.to_csv(
    os.path.join(
        PROJECT_DIR,
        "training_history.csv"
    ),
    index=False
)

# ============================================================
# 15. FINAL OUTPUT
# ============================================================

print("\n" + "=" * 60)

print("MODEL TRAINING COMPLETED SUCCESSFULLY")

print("=" * 60)

print("\nGenerated Files:")

for filename in os.listdir(PROJECT_DIR):

    if filename.endswith(
        (".h5", ".pkl", ".yaml", ".json", ".csv")
    ):

        print(filename)

print("\nAll files saved in:")
print(PROJECT_DIR)

print("\nIMPORTANT:")
print(
    "Metrics are calculated on training data only. "
    "They are not estimates of real-world predictive accuracy."
)
print(
    "The dataset is too small for reliable machine learning "
    "forecasting. Future values are experimental."
)
```
