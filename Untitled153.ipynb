```python
import os
import json
import yaml
import joblib
import h5py
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

np.random.seed(42)

# ============================================================
# 1. PROJECT CONFIGURATION
# ============================================================

PROJECT_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Government Fund Allocation and Utilization "
    r"Analysis and Forecasting"
)

CSV_PATH = os.path.join(
    PROJECT_DIR,
    "RS_Session_270_AU_2397_B.ii_.csv"
)

os.makedirs(PROJECT_DIR, exist_ok=True)


def output_path(filename):
    return os.path.join(
        PROJECT_DIR,
        "pso_" + filename
    )


def save_graph(filename):
    plt.tight_layout()
    plt.savefig(
        output_path(filename),
        dpi=300,
        bbox_inches="tight"
    )
    plt.show()
    plt.close()


# ============================================================
# 2. LOAD DATASET
# ============================================================

print("\nLoading Government Fund Dataset...")

df = pd.read_csv(CSV_PATH)

df.columns = df.columns.str.strip()

print("\nOriginal Dataset:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nAvailable Columns:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.isnull().sum())

# ============================================================
# 3. IDENTIFY DATASET COLUMNS
# ============================================================

year_col = next(
    (
        col for col in df.columns
        if "year" in col.lower()
    ),
    df.columns[0]
)

allocated_col = next(
    (
        col for col in df.columns
        if "allocat" in col.lower()
    ),
    None
)

utilized_col = next(
    (
        col for col in df.columns
        if (
            "utilis" in col.lower()
            or "utiliz" in col.lower()
        )
    ),
    None
)

if allocated_col is None or utilized_col is None:
    raise ValueError(
        "Allocation or utilization columns not found."
    )

print("\nDetected Columns:")
print("Year:", year_col)
print("Allocated:", allocated_col)
print("Utilized:", utilized_col)

# ============================================================
# 4. DATA PREPROCESSING
# ============================================================

df["Year_Numeric"] = pd.to_numeric(
    df[year_col]
    .astype(str)
    .str.extract(r"(\d{4})")[0],
    errors="coerce"
)

for col in [allocated_col, utilized_col]:

    df[col] = pd.to_numeric(
        df[col]
        .astype(str)
        .str.replace(",", "", regex=False),
        errors="coerce"
    )

train_df = df.dropna(
    subset=[
        "Year_Numeric",
        allocated_col,
        utilized_col
    ]
).copy()

train_df = train_df.sort_values(
    "Year_Numeric"
)

if len(train_df) < 3:
    raise ValueError(
        "At least three complete historical records are required."
    )

print("\nCleaned Dataset:")
print(train_df)

# ============================================================
# 5. FEATURE AND TARGET SELECTION
# ============================================================

X = train_df[
    ["Year_Numeric"]
].values.astype(float)

y = train_df[
    [allocated_col, utilized_col]
].values.astype(float)

years = X.flatten()

# ============================================================
# 6. FEATURE SCALING
# ============================================================

x_scaler = StandardScaler()
y_scaler = StandardScaler()

X_scaled = x_scaler.fit_transform(X)
y_scaled = y_scaler.fit_transform(y)

joblib.dump(
    {
        "x_scaler": x_scaler,
        "y_scaler": y_scaler,
        "target_columns": [
            allocated_col,
            utilized_col
        ]
    },
    output_path("scalers.pkl")
)

# ============================================================
# 7. PSO REGRESSION MODEL
# ============================================================

# Polynomial regression:
#
# y = w0 + w1*x + w2*x^2
#
# PSO optimizes the regression coefficients.
#
# Two targets:
# 1. Funds Allocated
# 2. Funds Utilized

POLYNOMIAL_DEGREE = 2


def polynomial_features(X_input):

    x = np.asarray(
        X_input,
        dtype=float
    ).reshape(-1)

    features = [
        np.ones_like(x)
    ]

    for degree in range(
        1,
        POLYNOMIAL_DEGREE + 1
    ):

        features.append(
            x ** degree
        )

    return np.column_stack(features)


X_poly = polynomial_features(X_scaled)

N_FEATURES = X_poly.shape[1]
N_TARGETS = y_scaled.shape[1]

N_PARAMETERS = N_FEATURES * N_TARGETS


def decode_particle(particle):

    return np.asarray(
        particle
    ).reshape(
        N_FEATURES,
        N_TARGETS
    )


def pso_predict_scaled(
    particle,
    X_input
):

    weights = decode_particle(
        particle
    )

    X_features = polynomial_features(
        X_input
    )

    return X_features @ weights


# ============================================================
# 8. PSO FITNESS FUNCTION
# ============================================================

REGULARIZATION = 0.001


def fitness_function(particle):

    predictions = pso_predict_scaled(
        particle,
        X_scaled
    )

    mse = np.mean(
        (
            y_scaled - predictions
        ) ** 2
    )

    regularization = (
        REGULARIZATION
        * np.sum(
            particle ** 2
        )
    )

    return float(
        mse + regularization
    )


# ============================================================
# 9. PSO CONFIGURATION
# ============================================================

NUM_PARTICLES = 50
MAX_ITERATIONS = 200

INERTIA_WEIGHT = 0.7

COGNITIVE_COEFFICIENT = 1.5

SOCIAL_COEFFICIENT = 1.5

LOWER_BOUND = -5.0
UPPER_BOUND = 5.0

MAX_VELOCITY = 1.0

# ============================================================
# 10. INITIALIZE SWARM
# ============================================================

print("\nInitializing PSO Swarm...")

positions = np.random.uniform(
    LOWER_BOUND,
    UPPER_BOUND,
    size=(
        NUM_PARTICLES,
        N_PARAMETERS
    )
)

velocities = np.random.uniform(
    -0.5,
    0.5,
    size=(
        NUM_PARTICLES,
        N_PARAMETERS
    )
)

# Personal best positions

personal_best_positions = positions.copy()

personal_best_fitness = np.array([

    fitness_function(particle)

    for particle in positions

])

# Global best position

best_index = np.argmin(
    personal_best_fitness
)

global_best_position = (
    personal_best_positions[
        best_index
    ].copy()
)

global_best_fitness = float(
    personal_best_fitness[
        best_index
    ]
)

# Optimization history

best_fitness_history = []
average_fitness_history = []

# ============================================================
# 11. PARTICLE SWARM OPTIMIZATION
# ============================================================

print("\nStarting PSO Optimization...")

for iteration in range(
    MAX_ITERATIONS
):

    # --------------------------------------------------------
    # Evaluate each particle
    # --------------------------------------------------------

    current_fitness = np.array([

        fitness_function(
            particle
        )

        for particle in positions

    ])

    # --------------------------------------------------------
    # Update personal best
    # --------------------------------------------------------

    improved = (
        current_fitness
        < personal_best_fitness
    )

    personal_best_positions[
        improved
    ] = positions[
        improved
    ].copy()

    personal_best_fitness[
        improved
    ] = current_fitness[
        improved
    ]

    # --------------------------------------------------------
    # Update global best
    # --------------------------------------------------------

    best_index = np.argmin(
        personal_best_fitness
    )

    if (
        personal_best_fitness[
            best_index
        ]
        < global_best_fitness
    ):

        global_best_fitness = float(
            personal_best_fitness[
                best_index
            ]
        )

        global_best_position = (
            personal_best_positions[
                best_index
            ].copy()
        )

    # --------------------------------------------------------
    # Record optimization history
    # --------------------------------------------------------

    best_fitness_history.append(
        global_best_fitness
    )

    average_fitness_history.append(
        float(
            np.mean(
                current_fitness
            )
        )
    )

    # --------------------------------------------------------
    # Update velocity and position
    # --------------------------------------------------------

    r1 = np.random.rand(
        NUM_PARTICLES,
        N_PARAMETERS
    )

    r2 = np.random.rand(
        NUM_PARTICLES,
        N_PARAMETERS
    )

    cognitive_component = (
        COGNITIVE_COEFFICIENT
        * r1
        * (
            personal_best_positions
            - positions
        )
    )

    social_component = (
        SOCIAL_COEFFICIENT
        * r2
        * (
            global_best_position
            - positions
        )
    )

    velocities = (
        INERTIA_WEIGHT
        * velocities
        + cognitive_component
        + social_component
    )

    velocities = np.clip(
        velocities,
        -MAX_VELOCITY,
        MAX_VELOCITY
    )

    positions = (
        positions + velocities
    )

    positions = np.clip(
        positions,
        LOWER_BOUND,
        UPPER_BOUND
    )

    if (
        iteration + 1
    ) % 20 == 0:

        print(
            f"Iteration {iteration + 1}"
            f" | Best Fitness: "
            f"{global_best_fitness:.6f}"
        )

# Evaluate final swarm positions as well

final_fitness = np.array([
    fitness_function(p)
    for p in positions
])

final_best_index = np.argmin(
    final_fitness
)

if (
    final_fitness[final_best_index]
    < global_best_fitness
):

    global_best_fitness = float(
        final_fitness[final_best_index]
    )

    global_best_position = (
        positions[final_best_index].copy()
    )

print("\nPSO Optimization Completed!")

print(
    "Best Fitness:",
    global_best_fitness
)

# ============================================================
# 12. FINAL PSO MODEL PREDICTIONS
# ============================================================

pso_predictions_scaled = (
    pso_predict_scaled(
        global_best_position,
        X_scaled
    )
)

pso_predictions = (
    y_scaler.inverse_transform(
        pso_predictions_scaled
    )
)

# ============================================================
# 13. MODEL EVALUATION
# ============================================================

mae = mean_absolute_error(
    y,
    pso_predictions
)

mse = mean_squared_error(
    y,
    pso_predictions
)

rmse = np.sqrt(mse)

r2 = r2_score(
    y,
    pso_predictions
)

allocated_mae = mean_absolute_error(
    y[:, 0],
    pso_predictions[:, 0]
)

utilized_mae = mean_absolute_error(
    y[:, 1],
    pso_predictions[:, 1]
)

allocated_rmse = np.sqrt(
    mean_squared_error(
        y[:, 0],
        pso_predictions[:, 0]
    )
)

utilized_rmse = np.sqrt(
    mean_squared_error(
        y[:, 1],
        pso_predictions[:, 1]
    )
)

allocated_r2 = r2_score(
    y[:, 0],
    pso_predictions[:, 0]
)

utilized_r2 = r2_score(
    y[:, 1],
    pso_predictions[:, 1]
)

print("\nPSO Model Training Metrics:")

print(f"MAE  : {mae:.6f}")
print(f"MSE  : {mse:.6f}")
print(f"RMSE : {rmse:.6f}")
print(f"R2   : {r2:.6f}")

# ============================================================
# 14. SAVE METRICS CSV
# ============================================================

metrics_df = pd.DataFrame({

    "Target": [
        "Allocated Funds",
        "Utilized Funds",
        "Overall"
    ],

    "MAE": [
        allocated_mae,
        utilized_mae,
        mae
    ],

    "RMSE": [
        allocated_rmse,
        utilized_rmse,
        rmse
    ],

    "R2_Score": [
        allocated_r2,
        utilized_r2,
        r2
    ]

})

metrics_df.to_csv(
    output_path(
        "model_comparison.csv"
    ),
    index=False
)

print("\nPSO Model Metrics:")
print(metrics_df)

# ============================================================
# 15. ACCURACY GRAPH
# ============================================================

# R2 represents regression training fit,
# not classification accuracy.

plt.figure(
    figsize=(10, 6)
)

bars = plt.bar(
    metrics_df["Target"],
    metrics_df["R2_Score"],
    color=[
        "royalblue",
        "darkorange",
        "seagreen"
    ]
)

plt.axhline(
    0,
    color="black",
    linewidth=0.8
)

plt.title(
    "PSO Model Training Fit (R²)",
    fontsize=14
)

plt.xlabel("Prediction Targets")
plt.ylabel("R² Score")

plt.grid(
    axis="y",
    alpha=0.3
)

for bar, value in zip(
    bars,
    metrics_df["R2_Score"]
):

    plt.text(
        bar.get_x()
        + bar.get_width() / 2,
        value,
        f"{value:.4f}",
        ha="center",
        va="bottom"
        if value >= 0
        else "top"
    )

save_graph(
    "accuracy_graph.png"
)

# ============================================================
# 16. COMPARISON GRAPH
# ============================================================

comparison_df = metrics_df.melt(
    id_vars="Target",
    value_vars=[
        "MAE",
        "RMSE"
    ],
    var_name="Metric",
    value_name="Value"
)

plt.figure(
    figsize=(11, 6)
)

sns.barplot(
    data=comparison_df,
    x="Target",
    y="Value",
    hue="Metric"
)

plt.title(
    "PSO Model Error Comparison"
)

plt.xlabel("Prediction Targets")
plt.ylabel("Error Value")

plt.grid(
    axis="y",
    alpha=0.3
)

save_graph(
    "comparison_graph.png"
)

# ============================================================
# 17. CORRELATION HEATMAP
# ============================================================

correlation_df = train_df[
    [
        "Year_Numeric",
        allocated_col,
        utilized_col
    ]
]

correlation_matrix = (
    correlation_df.corr()
)

plt.figure(
    figsize=(8, 6)
)

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5,
    vmin=-1,
    vmax=1
)

plt.title(
    "PSO Government Fund Correlation Heatmap"
)

save_graph(
    "heatmap.png"
)

# ============================================================
# 18. HISTORICAL RESULT CSV
# ============================================================

result_df = pd.DataFrame({

    "Year":
        years.astype(int),

    "Actual_Allocated":
        y[:, 0],

    "Predicted_Allocated":
        pso_predictions[:, 0],

    "Actual_Utilized":
        y[:, 1],

    "Predicted_Utilized":
        pso_predictions[:, 1]

})

result_df[
    "Allocated_Error"
] = (
    result_df["Actual_Allocated"]
    - result_df["Predicted_Allocated"]
)

result_df[
    "Utilized_Error"
] = (
    result_df["Actual_Utilized"]
    - result_df["Predicted_Utilized"]
)

result_df[
    "Allocated_Absolute_Error"
] = result_df[
    "Allocated_Error"
].abs()

result_df[
    "Utilized_Absolute_Error"
] = result_df[
    "Utilized_Error"
].abs()

result_df.to_csv(
    output_path(
        "result.csv"
    ),
    index=False
)

print("\nPSO Historical Results:")
print(result_df)

# ============================================================
# 19. HISTORICAL RESULT GRAPH
# ============================================================

plt.figure(
    figsize=(12, 6)
)

plt.plot(
    result_df["Year"],
    result_df["Actual_Allocated"],
    marker="o",
    linewidth=2,
    label="Actual Allocated"
)

plt.plot(
    result_df["Year"],
    result_df["Predicted_Allocated"],
    marker="o",
    linestyle="--",
    linewidth=2,
    label="PSO Predicted Allocated"
)

plt.plot(
    result_df["Year"],
    result_df["Actual_Utilized"],
    marker="s",
    linewidth=2,
    label="Actual Utilized"
)

plt.plot(
    result_df["Year"],
    result_df["Predicted_Utilized"],
    marker="s",
    linestyle="--",
    linewidth=2,
    label="PSO Predicted Utilized"
)

plt.title(
    "PSO Actual vs Predicted Government Funds"
)

plt.xlabel("Year")
plt.ylabel("Funds")

plt.legend()
plt.grid(alpha=0.3)

save_graph(
    "result_graph.png"
)

# ============================================================
# 20. FUTURE FUND FORECASTING
# ============================================================

last_year = int(
    years.max()
)

FORECAST_YEARS = 5

future_years = np.arange(
    last_year + 1,
    last_year + FORECAST_YEARS + 1
).reshape(-1, 1)

future_scaled = x_scaler.transform(
    future_years
)

future_predictions_scaled = (
    pso_predict_scaled(
        global_best_position,
        future_scaled
    )
)

future_predictions = (
    y_scaler.inverse_transform(
        future_predictions_scaled
    )
)

# ============================================================
# 21. FUTURE PREDICTION CSV
# ============================================================

prediction_df = pd.DataFrame({

    "Year":
        future_years.flatten().astype(int),

    "Predicted_Allocated":
        future_predictions[:, 0],

    "Predicted_Utilized":
        future_predictions[:, 1]

})

prediction_df[
    "Financial_Year"
] = (
    prediction_df["Year"]
    .astype(str)
    + "-"
    + (
        prediction_df["Year"] + 1
    ).astype(str).str[-2:]
)

prediction_df[
    "Predicted_Fund_Gap"
] = (
    prediction_df["Predicted_Allocated"]
    - prediction_df["Predicted_Utilized"]
)

prediction_df[
    "Predicted_Utilization_Percentage"
] = np.where(
    prediction_df["Predicted_Allocated"] != 0,
    (
        prediction_df["Predicted_Utilized"]
        / prediction_df["Predicted_Allocated"]
    ) * 100,
    np.nan
)

prediction_df.to_csv(
    output_path(
        "prediction.csv"
    ),
    index=False
)

print("\nPSO Future Predictions:")
print(prediction_df)

# ============================================================
# 22. FUTURE PREDICTION GRAPH
# ============================================================

plt.figure(
    figsize=(12, 6)
)

plt.plot(
    years,
    y[:, 0],
    marker="o",
    linewidth=2,
    label="Historical Allocated"
)

plt.plot(
    years,
    y[:, 1],
    marker="s",
    linewidth=2,
    label="Historical Utilized"
)

plt.plot(
    prediction_df["Year"],
    prediction_df["Predicted_Allocated"],
    marker="o",
    linestyle="--",
    linewidth=2,
    label="PSO Forecast Allocated"
)

plt.plot(
    prediction_df["Year"],
    prediction_df["Predicted_Utilized"],
    marker="s",
    linestyle="--",
    linewidth=2,
    label="PSO Forecast Utilized"
)

plt.axvline(
    last_year + 0.5,
    linestyle=":",
    color="gray",
    label="Forecast Boundary"
)

plt.title(
    "PSO Government Fund Allocation and Utilization Forecast"
)

plt.xlabel("Year")
plt.ylabel("Funds")

plt.legend()
plt.grid(alpha=0.3)

save_graph(
    "prediction_graph.png"
)

# ============================================================
# 23. PSO CONVERGENCE GRAPH
# ============================================================

optimization_df = pd.DataFrame({

    "Iteration":
        np.arange(
            1,
            MAX_ITERATIONS + 1
        ),

    "Best_Fitness":
        best_fitness_history,

    "Average_Fitness":
        average_fitness_history

})

optimization_df.to_csv(
    output_path(
        "optimization_history.csv"
    ),
    index=False
)

plt.figure(
    figsize=(12, 6)
)

plt.plot(
    optimization_df["Iteration"],
    optimization_df["Best_Fitness"],
    linewidth=2,
    label="Global Best Fitness"
)

plt.plot(
    optimization_df["Iteration"],
    optimization_df["Average_Fitness"],
    linestyle="--",
    label="Average Swarm Fitness"
)

plt.title(
    "PSO Optimization Convergence"
)

plt.xlabel("Iteration")
plt.ylabel("Fitness (Loss)")

plt.legend()
plt.grid(alpha=0.3)

save_graph(
    "convergence_graph.png"
)

# ============================================================
# 24. FUND UTILIZATION ANALYSIS
# ============================================================

utilization_df = pd.DataFrame({

    "Year":
        years.astype(int),

    "Funds_Allocated":
        y[:, 0],

    "Funds_Utilized":
        y[:, 1]

})

utilization_df[
    "Utilization_Percentage"
] = np.where(
    utilization_df["Funds_Allocated"] != 0,
    (
        utilization_df["Funds_Utilized"]
        / utilization_df["Funds_Allocated"]
    ) * 100,
    np.nan
)

utilization_df[
    "Fund_Gap"
] = (
    utilization_df["Funds_Allocated"]
    - utilization_df["Funds_Utilized"]
)

utilization_df.to_csv(
    output_path(
        "utilization_analysis.csv"
    ),
    index=False
)

plt.figure(
    figsize=(10, 6)
)

sns.barplot(
    data=utilization_df,
    x="Year",
    y="Utilization_Percentage",
    color="teal"
)

plt.axhline(
    100,
    linestyle="--",
    color="red",
    label="100% Utilization"
)

plt.title(
    "PSO Government Fund Utilization Analysis"
)

plt.xlabel("Year")
plt.ylabel("Utilization Percentage (%)")

plt.legend()
plt.grid(
    axis="y",
    alpha=0.3
)

save_graph(
    "utilization_graph.png"
)

# ============================================================
# 25. ACTUAL VS PREDICTED SCATTER GRAPH
# ============================================================

plt.figure(
    figsize=(10, 6)
)

plt.scatter(
    y[:, 0],
    pso_predictions[:, 0],
    s=90,
    label="Allocated Funds"
)

plt.scatter(
    y[:, 1],
    pso_predictions[:, 1],
    s=90,
    label="Utilized Funds"
)

all_values = np.concatenate(
    [
        y.flatten(),
        pso_predictions.flatten()
    ]
)

minimum = np.min(all_values)
maximum = np.max(all_values)

plt.plot(
    [minimum, maximum],
    [minimum, maximum],
    linestyle="--",
    color="black",
    label="Perfect Prediction"
)

plt.title(
    "PSO Actual vs Predicted Scatter Plot"
)

plt.xlabel("Actual Values")
plt.ylabel("Predicted Values")

plt.legend()
plt.grid(alpha=0.3)

save_graph(
    "scatter_graph.png"
)

# ============================================================
# 26. SAVE PSO MODEL AS PKL
# ============================================================

pso_model_data = {

    "algorithm":
        "Particle Swarm Optimization",

    "model_type":
        "Polynomial Regression",

    "best_particle":
        global_best_position,

    "polynomial_degree":
        POLYNOMIAL_DEGREE,

    "n_features":
        N_FEATURES,

    "n_targets":
        N_TARGETS,

    "x_scaler":
        x_scaler,

    "y_scaler":
        y_scaler,

    "target_columns":
        [
            allocated_col,
            utilized_col
        ],

    "best_fitness":
        global_best_fitness

}

joblib.dump(
    pso_model_data,
    output_path(
        "model.pkl"
    )
)

print("\nPSO PKL Model Saved!")

# ============================================================
# 27. SAVE PSO MODEL AS H5
# ============================================================

# This H5 file stores PSO-optimized regression
# coefficients and preprocessing parameters.
#
# It is not a Keras neural network model.

with h5py.File(
    output_path(
        "model.h5"
    ),
    "w"
) as h5_file:

    h5_file.create_dataset(
        "best_particle",
        data=global_best_position
    )

    h5_file.create_dataset(
        "weights",
        data=decode_particle(
            global_best_position
        )
    )

    h5_file.create_dataset(
        "x_mean",
        data=x_scaler.mean_
    )

    h5_file.create_dataset(
        "x_scale",
        data=x_scaler.scale_
    )

    h5_file.create_dataset(
        "y_mean",
        data=y_scaler.mean_
    )

    h5_file.create_dataset(
        "y_scale",
        data=y_scaler.scale_
    )

    h5_file.attrs[
        "algorithm"
    ] = "Particle Swarm Optimization"

    h5_file.attrs[
        "polynomial_degree"
    ] = POLYNOMIAL_DEGREE

    h5_file.attrs[
        "best_fitness"
    ] = global_best_fitness

print("\nPSO H5 Model Saved!")

# ============================================================
# 28. SAVE YAML CONFIGURATION
# ============================================================

config = {

    "project":
        "AI-Based Government Fund Allocation and Utilization Analysis and Forecasting",

    "algorithm":
        "Particle Swarm Optimization",

    "dataset":
        os.path.basename(
            CSV_PATH
        ),

    "total_records":
        int(len(df)),

    "training_records":
        int(len(train_df)),

    "features":
        ["Year_Numeric"],

    "targets":
        [
            allocated_col,
            utilized_col
        ],

    "model": {

        "type":
            "Polynomial Regression",

        "degree":
            POLYNOMIAL_DEGREE,

        "parameters":
            N_PARAMETERS

    },

    "pso_parameters": {

        "particles":
            NUM_PARTICLES,

        "iterations":
            MAX_ITERATIONS,

        "inertia_weight":
            INERTIA_WEIGHT,

        "cognitive_coefficient":
            COGNITIVE_COEFFICIENT,

        "social_coefficient":
            SOCIAL_COEFFICIENT,

        "lower_bound":
            LOWER_BOUND,

        "upper_bound":
            UPPER_BOUND,

        "max_velocity":
            MAX_VELOCITY,

        "regularization":
            REGULARIZATION

    },

    "evaluation":
        "Training-fit metrics only",

    "forecast_reliability":
        "Experimental - insufficient historical data"

}

with open(
    output_path(
        "config.yaml"
    ),
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        config,
        file,
        sort_keys=False
    )

print("\nPSO YAML Configuration Saved!")

# ============================================================
# 29. SAVE JSON RESULTS
# ============================================================

json_results = {

    "project":
        "Government Fund Allocation and Utilization Forecasting",

    "algorithm":
        "Particle Swarm Optimization",

    "dataset_summary": {

        "total_records":
            int(len(df)),

        "training_records":
            int(len(train_df)),

        "missing_records":
            int(
                len(df) - len(train_df)
            )

    },

    "model_metrics": {

        "MAE":
            float(mae),

        "MSE":
            float(mse),

        "RMSE":
            float(rmse),

        "R2":
            float(r2),

        "Allocated_MAE":
            float(allocated_mae),

        "Utilized_MAE":
            float(utilized_mae),

        "Allocated_RMSE":
            float(allocated_rmse),

        "Utilized_RMSE":
            float(utilized_rmse),

        "Allocated_R2":
            float(allocated_r2),

        "Utilized_R2":
            float(utilized_r2)

    },

    "optimization": {

        "best_fitness":
            float(global_best_fitness),

        "iterations":
            MAX_ITERATIONS,

        "particles":
            NUM_PARTICLES

    },

    "evaluation_type":
        "In-sample training evaluation",

    "forecast_note":
        "Experimental predictions due to limited data",

    "future_predictions":
        json.loads(
            prediction_df.to_json(
                orient="records"
            )
        )

}

with open(
    output_path(
        "results.json"
    ),
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        json_results,
        file,
        indent=4
    )

print("\nPSO JSON Results Saved!")

# ============================================================
# 30. FINAL OUTPUT SUMMARY
# ============================================================

print("\n" + "=" * 65)

print(
    "PARTICLE SWARM OPTIMIZATION ANALYSIS COMPLETED"
)

print("=" * 65)

generated_files = [

    "accuracy_graph.png",
    "comparison_graph.png",
    "heatmap.png",
    "result.csv",
    "result_graph.png",
    "prediction.csv",
    "prediction_graph.png",
    "convergence_graph.png",
    "utilization_graph.png",
    "utilization_analysis.csv",
    "scatter_graph.png",
    "optimization_history.csv",
    "model_comparison.csv",
    "model.pkl",
    "model.h5",
    "scalers.pkl",
    "config.yaml",
    "results.json"

]

print("\nGenerated Files:\n")

for filename in generated_files:

    full_path = output_path(
        filename
    )

    if os.path.exists(
        full_path
    ):

        print(
            "[SAVED] pso_" + filename
        )

    else:

        print(
            "[MISSING] pso_" + filename
        )

print("\nOutput Directory:")
print(PROJECT_DIR)

print("\nPSO Training Metrics:")

print(f"MAE  : {mae:.6f}")
print(f"MSE  : {mse:.6f}")
print(f"RMSE : {rmse:.6f}")
print(f"R2   : {r2:.6f}")

print("\nBest PSO Fitness:")
print(global_best_fitness)

print("\nIMPORTANT:")
print(
    "These metrics measure training fit only, "
    "not independently validated accuracy."
)

print(
    "Future predictions are experimental "
    "because the dataset is extremely small."
)

print("\nPROGRAM EXECUTION COMPLETED SUCCESSFULLY!")
```
