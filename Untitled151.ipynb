```python
import os
import json
import yaml
import joblib
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf

from sklearn.linear_model import LinearRegression
from sklearn.ensemble import (
    RandomForestRegressor,
    GradientBoostingRegressor
)
from sklearn.tree import DecisionTreeRegressor
from sklearn.svm import SVR
from sklearn.multioutput import MultiOutputRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

np.random.seed(42)
tf.random.set_seed(42)

# ============================================================
# 1. PROJECT PATH
# ============================================================

PROJECT_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Government Fund Allocation and Utilization "
    r"Analysis and Forecasting"
)

CSV_PATH = os.path.join(
    PROJECT_DIR,
    "RS_Session_270_AU_2397_B.ii_.csv"
)

os.makedirs(PROJECT_DIR, exist_ok=True)

# ============================================================
# 2. LOAD DATASET
# ============================================================

print("\nLoading dataset...")

df = pd.read_csv(CSV_PATH)

df.columns = df.columns.str.strip()

print("\nOriginal Dataset:")
print(df.head())

print("\nDataset Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.isnull().sum())

# ============================================================
# 3. IDENTIFY COLUMNS
# ============================================================

year_col = next(
    (
        col for col in df.columns
        if "year" in col.lower()
    ),
    df.columns[0]
)

allocated_col = next(
    (
        col for col in df.columns
        if "allocat" in col.lower()
    ),
    None
)

utilized_col = next(
    (
        col for col in df.columns
        if (
            "utilis" in col.lower()
            or "utiliz" in col.lower()
        )
    ),
    None
)

if allocated_col is None or utilized_col is None:
    raise ValueError(
        "Fund allocation/utilization columns not found."
    )

print("\nDetected Columns:")
print("Year:", year_col)
print("Allocated:", allocated_col)
print("Utilized:", utilized_col)

# ============================================================
# 4. DATA PREPROCESSING
# ============================================================

df["Year_Numeric"] = pd.to_numeric(
    df[year_col]
    .astype(str)
    .str.extract(r"(\d{4})")[0],
    errors="coerce"
)

for col in [allocated_col, utilized_col]:

    df[col] = pd.to_numeric(
        df[col]
        .astype(str)
        .str.replace(",", "", regex=False),
        errors="coerce"
    )

train_df = df.dropna(
    subset=[
        "Year_Numeric",
        allocated_col,
        utilized_col
    ]
).copy()

train_df = train_df.sort_values("Year_Numeric")

if len(train_df) < 3:
    raise ValueError(
        "At least 3 complete records are required."
    )

print("\nCleaned Dataset:")
print(train_df)

# ============================================================
# 5. FEATURE AND TARGET SELECTION
# ============================================================

X = train_df[["Year_Numeric"]].values

y = train_df[
    [allocated_col, utilized_col]
].values

# ============================================================
# 6. DATA SCALING
# ============================================================

x_scaler = StandardScaler()
y_scaler = StandardScaler()

X_scaled = x_scaler.fit_transform(X)
y_scaled = y_scaler.fit_transform(y)

joblib.dump(
    {
        "x_scaler": x_scaler,
        "y_scaler": y_scaler
    },
    os.path.join(PROJECT_DIR, "scalers.pkl")
)

# ============================================================
# 7. MACHINE LEARNING MODELS
# ============================================================

models = {

    "Linear Regression":
        LinearRegression(),

    "Random Forest":
        RandomForestRegressor(
            n_estimators=100,
            random_state=42
        ),

    "Decision Tree":
        DecisionTreeRegressor(
            random_state=42
        ),

    "Gradient Boosting":
        MultiOutputRegressor(
            GradientBoostingRegressor(
                n_estimators=100,
                random_state=42
            )
        ),

    "Support Vector Regression":
        MultiOutputRegressor(
            SVR(
                kernel="rbf",
                C=100
            )
        )
}

# ============================================================
# 8. TRAIN AND EVALUATE MODELS
# ============================================================

results = []
predictions = {}

print("\nTraining Machine Learning Models...")

for name, model in models.items():

    print("\nTraining:", name)

    # SVR uses standardized inputs and targets
    if name == "Support Vector Regression":

        model.fit(X_scaled, y_scaled)

        pred_scaled = model.predict(X_scaled)

        pred = y_scaler.inverse_transform(
            pred_scaled
        )

    else:

        model.fit(X, y)

        pred = model.predict(X)

    predictions[name] = pred

    mae = mean_absolute_error(y, pred)

    mse = mean_squared_error(y, pred)

    rmse = np.sqrt(mse)

    r2 = r2_score(y, pred)

    results.append({

        "Model": name,
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R2_Score": r2

    })

    print("MAE:", mae)
    print("MSE:", mse)
    print("RMSE:", rmse)
    print("R2:", r2)

# ============================================================
# 9. NEURAL NETWORK
# ============================================================

print("\nTraining Neural Network...")

nn_model = tf.keras.Sequential([

    tf.keras.layers.Input(shape=(1,)),

    tf.keras.layers.Dense(
        32,
        activation="relu"
    ),

    tf.keras.layers.Dense(
        16,
        activation="relu"
    ),

    tf.keras.layers.Dense(
        2,
        activation="linear"
    )

])

nn_model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.01
    ),
    loss="mse",
    metrics=["mae"]
)

history = nn_model.fit(
    X_scaled,
    y_scaled,
    epochs=200,
    batch_size=2,
    verbose=0
)

nn_pred_scaled = nn_model.predict(
    X_scaled,
    verbose=0
)

nn_pred = y_scaler.inverse_transform(
    nn_pred_scaled
)

predictions["Neural Network"] = nn_pred

nn_mae = mean_absolute_error(y, nn_pred)
nn_mse = mean_squared_error(y, nn_pred)
nn_rmse = np.sqrt(nn_mse)
nn_r2 = r2_score(y, nn_pred)

results.append({

    "Model": "Neural Network",
    "MAE": nn_mae,
    "MSE": nn_mse,
    "RMSE": nn_rmse,
    "R2_Score": nn_r2

})

# Save models
joblib.dump(
    models["Random Forest"],
    os.path.join(
        PROJECT_DIR,
        "government_fund_model.pkl"
    )
)

nn_model.save(
    os.path.join(
        PROJECT_DIR,
        "government_fund_model.h5"
    )
)

# ============================================================
# 10. MODEL COMPARISON DATA
# ============================================================

results_df = pd.DataFrame(results)

results_df.to_csv(
    os.path.join(
        PROJECT_DIR,
        "model_comparison.csv"
    ),
    index=False
)

print("\nModel Comparison:")
print(results_df)

# ============================================================
# 11. ACCURACY GRAPH
# ============================================================

# R2 is a regression fit score, not classification accuracy.

plt.figure(figsize=(12, 6))

sns.barplot(
    data=results_df,
    x="Model",
    y="R2_Score",
    color="steelblue"
)

plt.title(
    "Machine Learning Model Training Fit (R²)",
    fontsize=14
)

plt.xlabel("Machine Learning Models")
plt.ylabel("Training R² Score")

plt.xticks(rotation=30, ha="right")

plt.grid(axis="y", alpha=0.3)

plt.tight_layout()

plt.savefig(
    os.path.join(
        PROJECT_DIR,
        "accuracy_graph.png"
    ),
    dpi=300
)

plt.show()

# ============================================================
# 12. MODEL COMPARISON GRAPH
# ============================================================

comparison_melted = results_df.melt(
    id_vars="Model",
    value_vars=["MAE", "RMSE"],
    var_name="Metric",
    value_name="Value"
)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=comparison_melted,
    x="Model",
    y="Value",
    hue="Metric"
)

plt.title(
    "Machine Learning Model Comparison"
)

plt.xlabel("Models")
plt.ylabel("Error Value")

plt.xticks(rotation=30, ha="right")

plt.grid(axis="y", alpha=0.3)

plt.tight_layout()

plt.savefig(
    os.path.join(
        PROJECT_DIR,
        "comparison_graph.png"
    ),
    dpi=300
)

plt.show()

# ============================================================
# 13. CORRELATION HEATMAP
# ============================================================

heatmap_df = train_df[
    [
        "Year_Numeric",
        allocated_col,
        utilized_col
    ]
]

correlation = heatmap_df.corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title(
    "Government Fund Dataset Correlation Heatmap"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        PROJECT_DIR,
        "heatmap.png"
    ),
    dpi=300
)

plt.show()

# ============================================================
# 14. HISTORICAL RESULT CSV
# ============================================================

result_df = pd.DataFrame({

    "Year": train_df["Year_Numeric"].values,

    "Actual_Allocated":
        y[:, 0],

    "Actual_Utilized":
        y[:, 1]

})

for name, pred in predictions.items():

    safe_name = (
        name.replace(" ", "_")
    )

    result_df[
        safe_name + "_Allocated"
    ] = pred[:, 0]

    result_df[
        safe_name + "_Utilized"
    ] = pred[:, 1]

result_df.to_csv(
    os.path.join(
        PROJECT_DIR,
        "result.csv"
    ),
    index=False
)

print("\nHistorical Results:")
print(result_df)

# ============================================================
# 15. HISTORICAL RESULT GRAPH
# ============================================================

plt.figure(figsize=(12, 6))

plt.plot(
    result_df["Year"],
    result_df["Actual_Allocated"],
    marker="o",
    linewidth=2,
    label="Actual Allocated"
)

plt.plot(
    result_df["Year"],
    result_df["Actual_Utilized"],
    marker="s",
    linewidth=2,
    label="Actual Utilized"
)

plt.plot(
    result_df["Year"],
    result_df["Random_Forest_Allocated"],
    linestyle="--",
    marker="o",
    label="RF Predicted Allocated"
)

plt.plot(
    result_df["Year"],
    result_df["Random_Forest_Utilized"],
    linestyle="--",
    marker="s",
    label="RF Predicted Utilized"
)

plt.title(
    "Actual vs Predicted Government Funds"
)

plt.xlabel("Year")
plt.ylabel("Funds")

plt.legend()

plt.grid(alpha=0.3)

plt.tight_layout()

plt.savefig(
    os.path.join(
        PROJECT_DIR,
        "result_graph.png"
    ),
    dpi=300
)

plt.show()

# ============================================================
# 16. FUTURE YEAR GENERATION
# ============================================================

last_year = int(
    train_df["Year_Numeric"].max()
)

# Include missing years such as 2024-25
# and continue forecasting through five years ahead.

future_years = np.arange(
    last_year + 1,
    last_year + 6
).reshape(-1, 1)

future_scaled = x_scaler.transform(
    future_years
)

# ============================================================
# 17. FUTURE PREDICTIONS
# ============================================================

prediction_df = pd.DataFrame({

    "Year": future_years.flatten()

})

for name, model in models.items():

    if name == "Support Vector Regression":

        future_pred_scaled = model.predict(
            future_scaled
        )

        future_pred = y_scaler.inverse_transform(
            future_pred_scaled
        )

    else:

        future_pred = model.predict(
            future_years
        )

    safe_name = name.replace(" ", "_")

    prediction_df[
        safe_name + "_Allocated"
    ] = future_pred[:, 0]

    prediction_df[
        safe_name + "_Utilized"
    ] = future_pred[:, 1]

# Neural Network future predictions

nn_future_scaled = nn_model.predict(
    future_scaled,
    verbose=0
)

nn_future = y_scaler.inverse_transform(
    nn_future_scaled
)

prediction_df[
    "Neural_Network_Allocated"
] = nn_future[:, 0]

prediction_df[
    "Neural_Network_Utilized"
] = nn_future[:, 1]

# Financial year labels
prediction_df["Financial_Year"] = (
    prediction_df["Year"].astype(int).astype(str)
    + "-"
    + (
        prediction_df["Year"].astype(int) + 1
    ).astype(str).str[-2:]
)

prediction_df.to_csv(
    os.path.join(
        PROJECT_DIR,
        "prediction.csv"
    ),
    index=False
)

print("\nFuture Predictions:")
print(prediction_df)

# ============================================================
# 18. FUTURE PREDICTION GRAPH
# ============================================================

plt.figure(figsize=(12, 6))

plt.plot(
    train_df["Year_Numeric"],
    train_df[allocated_col],
    marker="o",
    linewidth=2,
    label="Historical Allocated"
)

plt.plot(
    train_df["Year_Numeric"],
    train_df[utilized_col],
    marker="s",
    linewidth=2,
    label="Historical Utilized"
)

plt.plot(
    prediction_df["Year"],
    prediction_df["Linear_Regression_Allocated"],
    marker="o",
    linestyle="--",
    label="Forecast Allocated (Linear Regression)"
)

plt.plot(
    prediction_df["Year"],
    prediction_df["Linear_Regression_Utilized"],
    marker="s",
    linestyle="--",
    label="Forecast Utilized (Linear Regression)"
)

plt.title(
    "Government Fund Allocation and Utilization Forecast"
)

plt.xlabel("Year")
plt.ylabel("Funds")

plt.legend()

plt.grid(alpha=0.3)

plt.tight_layout()

plt.savefig(
    os.path.join(
        PROJECT_DIR,
        "prediction_graph.png"
    ),
    dpi=300
)

plt.show()

# ============================================================
# 19. NEURAL NETWORK TRAINING LOSS GRAPH
# ============================================================

plt.figure(figsize=(10, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.title(
    "Neural Network Training Loss"
)

plt.xlabel("Epochs")
plt.ylabel("Mean Squared Error (Scaled)")

plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()

plt.savefig(
    os.path.join(
        PROJECT_DIR,
        "training_loss_graph.png"
    ),
    dpi=300
)

plt.show()

# ============================================================
# 20. FUND UTILIZATION ANALYSIS
# ============================================================

utilization_df = train_df[
    [
        "Year_Numeric",
        allocated_col,
        utilized_col
    ]
].copy()

utilization_df["Utilization_Percentage"] = np.where(
    utilization_df[allocated_col] != 0,
    (
        utilization_df[utilized_col]
        / utilization_df[allocated_col]
    ) * 100,
    np.nan
)

utilization_df.to_csv(
    os.path.join(
        PROJECT_DIR,
        "fund_utilization_analysis.csv"
    ),
    index=False
)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=utilization_df,
    x="Year_Numeric",
    y="Utilization_Percentage",
    color="teal"
)

plt.axhline(
    100,
    linestyle="--",
    label="100% Utilization"
)

plt.title(
    "Year-wise Government Fund Utilization"
)

plt.xlabel("Year")
plt.ylabel("Utilization (%)")

plt.legend()
plt.tight_layout()

plt.savefig(
    os.path.join(
        PROJECT_DIR,
        "fund_utilization_graph.png"
    ),
    dpi=300
)

plt.show()

# ============================================================
# 21. SAVE YAML CONFIGURATION
# ============================================================

config = {

    "project":
        "AI-Based Government Fund Allocation and Utilization Analysis and Forecasting",

    "dataset":
        os.path.basename(CSV_PATH),

    "total_records":
        int(len(df)),

    "training_records":
        int(len(train_df)),

    "features":
        ["Year_Numeric"],

    "targets":
        [allocated_col, utilized_col],

    "models":
        list(models.keys()) + ["Neural Network"],

    "evaluation_type":
        "In-sample training evaluation only",

    "forecast_note":
        "Experimental forecasts due to limited historical data"

}

with open(
    os.path.join(
        PROJECT_DIR,
        "model_config.yaml"
    ),
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        config,
        file,
        sort_keys=False
    )

# ============================================================
# 22. SAVE JSON RESULTS
# ============================================================

json_results = {

    "project":
        "Government Fund Allocation and Utilization Forecasting",

    "dataset_records":
        int(len(df)),

    "training_records":
        int(len(train_df)),

    "evaluation_type":
        "Training-fit metrics; no independent test set",

    "model_metrics":
        results_df.to_dict(
            orient="records"
        ),

    "future_predictions":
        prediction_df.to_dict(
            orient="records"
        )

}

with open(
    os.path.join(
        PROJECT_DIR,
        "model_results.json"
    ),
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        json_results,
        file,
        indent=4
    )

# ============================================================
# 23. SAVE NEURAL NETWORK TRAINING HISTORY
# ============================================================

pd.DataFrame(
    history.history
).to_csv(
    os.path.join(
        PROJECT_DIR,
        "training_history.csv"
    ),
    index=False
)

# ============================================================
# 24. FINAL OUTPUT
# ============================================================

print("\n" + "=" * 65)

print(
    "AI-BASED GOVERNMENT FUND ANALYSIS COMPLETED"
)

print("=" * 65)

print("\nGenerated Files:\n")

generated_files = [

    "accuracy_graph.png",
    "comparison_graph.png",
    "heatmap.png",
    "result.csv",
    "result_graph.png",
    "prediction.csv",
    "prediction_graph.png",
    "training_loss_graph.png",
    "fund_utilization_analysis.csv",
    "fund_utilization_graph.png",
    "model_comparison.csv",
    "government_fund_model.pkl",
    "government_fund_model.h5",
    "scalers.pkl",
    "model_config.yaml",
    "model_results.json",
    "training_history.csv"

]

for filename in generated_files:

    path = os.path.join(
        PROJECT_DIR,
        filename
    )

    if os.path.exists(path):
        print("[SAVED]", filename)
    else:
        print("[MISSING]", filename)

print("\nOutput Directory:")
print(PROJECT_DIR)

print("\nNOTE:")
print(
    "All model scores are training-fit scores, "
    "not validated prediction accuracy."
)

print(
    "Forecasts are experimental because "
    "the historical dataset is extremely small."
)

print("\nPROGRAM EXECUTION COMPLETED!")
```
