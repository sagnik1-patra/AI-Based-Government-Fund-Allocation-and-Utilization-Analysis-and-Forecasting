```python
import os
import json
import yaml
import joblib
import h5py
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

np.random.seed(42)

# ============================================================
# 1. PROJECT CONFIGURATION
# ============================================================

PROJECT_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Government Fund Allocation and Utilization "
    r"Analysis and Forecasting"
)

CSV_PATH = os.path.join(
    PROJECT_DIR,
    "RS_Session_270_AU_2397_B.ii_.csv"
)

os.makedirs(PROJECT_DIR, exist_ok=True)

def output_path(filename):
    return os.path.join(
        PROJECT_DIR,
        "ais_" + filename
    )

# ============================================================
# 2. LOAD DATASET
# ============================================================

print("\nLoading Government Fund Dataset...")

df = pd.read_csv(CSV_PATH)

df.columns = df.columns.str.strip()

print("\nOriginal Dataset:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nAvailable Columns:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.isnull().sum())

# ============================================================
# 3. IDENTIFY DATASET COLUMNS
# ============================================================

year_col = next(
    (
        col for col in df.columns
        if "year" in col.lower()
    ),
    df.columns[0]
)

allocated_col = next(
    (
        col for col in df.columns
        if "allocat" in col.lower()
    ),
    None
)

utilized_col = next(
    (
        col for col in df.columns
        if (
            "utilis" in col.lower()
            or "utiliz" in col.lower()
        )
    ),
    None
)

if allocated_col is None or utilized_col is None:
    raise ValueError(
        "Allocation or utilization columns not found."
    )

print("\nDetected Columns:")
print("Year:", year_col)
print("Allocated:", allocated_col)
print("Utilized:", utilized_col)

# ============================================================
# 4. DATA PREPROCESSING
# ============================================================

df["Year_Numeric"] = pd.to_numeric(
    df[year_col]
    .astype(str)
    .str.extract(r"(\d{4})")[0],
    errors="coerce"
)

for col in [allocated_col, utilized_col]:

    df[col] = pd.to_numeric(
        df[col]
        .astype(str)
        .str.replace(",", "", regex=False),
        errors="coerce"
    )

train_df = df.dropna(
    subset=[
        "Year_Numeric",
        allocated_col,
        utilized_col
    ]
).copy()

train_df = train_df.sort_values(
    "Year_Numeric"
)

if len(train_df) < 3:
    raise ValueError(
        "At least three complete historical records are required."
    )

print("\nCleaned Dataset:")
print(train_df)

# ============================================================
# 5. FEATURE AND TARGET SELECTION
# ============================================================

X = train_df[
    ["Year_Numeric"]
].values.astype(float)

y = train_df[
    [allocated_col, utilized_col]
].values.astype(float)

years = X.flatten()

# ============================================================
# 6. FEATURE SCALING
# ============================================================

x_scaler = StandardScaler()
y_scaler = StandardScaler()

X_scaled = x_scaler.fit_transform(X)

y_scaled = y_scaler.fit_transform(y)

joblib.dump(
    {
        "x_scaler": x_scaler,
        "y_scaler": y_scaler
    },
    output_path("scalers.pkl")
)

# ============================================================
# 7. AIS REGRESSION MODEL
# ============================================================

# Polynomial regression model:
#
# y = w0 + w1*x + w2*x^2
#
# AIS optimizes the regression coefficients.
#
# Two target variables:
# 1. Funds Allocated
# 2. Funds Utilized
#
# Each antibody contains six coefficients.

POLYNOMIAL_DEGREE = 2

def polynomial_features(X):

    x = np.asarray(X).reshape(-1)

    features = [
        np.ones_like(x)
    ]

    for degree in range(
        1,
        POLYNOMIAL_DEGREE + 1
    ):

        features.append(
            x ** degree
        )

    return np.column_stack(features)


X_poly = polynomial_features(
    X_scaled
)

N_FEATURES = X_poly.shape[1]

N_TARGETS = y_scaled.shape[1]

N_PARAMETERS = N_FEATURES * N_TARGETS

# ============================================================
# 8. ANTIBODY REPRESENTATION
# ============================================================

def decode_antibody(antibody):

    return antibody.reshape(
        N_FEATURES,
        N_TARGETS
    )


def ais_predict_scaled(
    antibody,
    X_input
):

    weights = decode_antibody(
        antibody
    )

    X_features = polynomial_features(
        X_input
    )

    return X_features @ weights


def ais_fitness(
    antibody,
    X_input,
    y_actual
):

    y_predicted = ais_predict_scaled(
        antibody,
        X_input
    )

    mse = np.mean(
        (
            y_actual - y_predicted
        ) ** 2
    )

    # Small regularization penalty
    regularization = (
        0.001 * np.sum(antibody ** 2)
    )

    return mse + regularization


# ============================================================
# 9. AIS CONFIGURATION
# ============================================================

POPULATION_SIZE = 50

GENERATIONS = 200

SELECTION_SIZE = 10

CLONE_FACTOR = 5

MUTATION_RATE = 0.15

RANDOM_REPLACEMENT = 5

LOWER_BOUND = -5

UPPER_BOUND = 5

# ============================================================
# 10. INITIALIZE ANTIBODY POPULATION
# ============================================================

population = np.random.uniform(
    LOWER_BOUND,
    UPPER_BOUND,
    size=(
        POPULATION_SIZE,
        N_PARAMETERS
    )
)

best_antibody = None

best_fitness = float("inf")

fitness_history = []

average_fitness_history = []

# ============================================================
# 11. AIS CLONAL SELECTION ALGORITHM
# ============================================================

print("\nStarting AIS Optimization...")

for generation in range(GENERATIONS):

    # --------------------------------------------------------
    # Calculate antibody fitness
    # --------------------------------------------------------

    fitness_values = np.array([

        ais_fitness(
            antibody,
            X_scaled,
            y_scaled
        )

        for antibody in population

    ])

    # --------------------------------------------------------
    # Sort antibodies by fitness
    # --------------------------------------------------------

    sorted_indices = np.argsort(
        fitness_values
    )

    population = population[
        sorted_indices
    ]

    fitness_values = fitness_values[
        sorted_indices
    ]

    # --------------------------------------------------------
    # Update global best antibody
    # --------------------------------------------------------

    if fitness_values[0] < best_fitness:

        best_fitness = float(
            fitness_values[0]
        )

        best_antibody = population[
            0
        ].copy()

    # --------------------------------------------------------
    # Select best antibodies
    # --------------------------------------------------------

    selected_antibodies = population[
        :SELECTION_SIZE
    ]

    selected_fitness = fitness_values[
        :SELECTION_SIZE
    ]

    # --------------------------------------------------------
    # Clone and mutate antibodies
    # --------------------------------------------------------

    clones = []

    for i, antibody in enumerate(
        selected_antibodies
    ):

        # Better antibodies produce more clones
        number_of_clones = max(
            1,
            int(
                CLONE_FACTOR
                * SELECTION_SIZE
                / (i + 1)
            )
        )

        for _ in range(
            number_of_clones
        ):

            clone = antibody.copy()

            # Better fitness -> smaller mutation
            mutation_strength = (
                MUTATION_RATE
                * (
                    0.1
                    + selected_fitness[i]
                )
            )

            mutation_strength = min(
                mutation_strength,
                1.0
            )

            mutation = np.random.normal(
                0,
                mutation_strength,
                size=N_PARAMETERS
            )

            clone += mutation

            clone = np.clip(
                clone,
                LOWER_BOUND,
                UPPER_BOUND
            )

            clones.append(
                clone
            )

    clones = np.array(
        clones
    )

    # --------------------------------------------------------
    # Combine population and clones
    # --------------------------------------------------------

    combined_population = np.vstack(
        [
            population,
            clones
        ]
    )

    combined_fitness = np.array([

        ais_fitness(
            antibody,
            X_scaled,
            y_scaled
        )

        for antibody in combined_population

    ])

    best_indices = np.argsort(
        combined_fitness
    )

    # Keep the strongest antibodies
    population = combined_population[
        best_indices[
            :POPULATION_SIZE - RANDOM_REPLACEMENT
        ]
    ]

    # --------------------------------------------------------
    # Introduce random antibodies
    # --------------------------------------------------------

    random_antibodies = np.random.uniform(
        LOWER_BOUND,
        UPPER_BOUND,
        size=(
            RANDOM_REPLACEMENT,
            N_PARAMETERS
        )
    )

    population = np.vstack(
        [
            population,
            random_antibodies
        ]
    )

    # --------------------------------------------------------
    # Track optimization history
    # --------------------------------------------------------

    fitness_history.append(
        best_fitness
    )

    average_fitness_history.append(
        float(
            np.mean(
                combined_fitness
            )
        )
    )

    if (
        generation + 1
    ) % 20 == 0:

        print(
            f"Generation {generation + 1}"
            f" | Best Fitness: {best_fitness:.6f}"
        )

print("\nAIS Optimization Completed!")

print(
    "Best Fitness:",
    best_fitness
)

# ============================================================
# 12. FINAL AIS MODEL PREDICTIONS
# ============================================================

ais_predictions_scaled = ais_predict_scaled(
    best_antibody,
    X_scaled
)

ais_predictions = y_scaler.inverse_transform(
    ais_predictions_scaled
)

# ============================================================
# 13. MODEL EVALUATION
# ============================================================

mae = mean_absolute_error(
    y,
    ais_predictions
)

mse = mean_squared_error(
    y,
    ais_predictions
)

rmse = np.sqrt(mse)

r2 = r2_score(
    y,
    ais_predictions
)

print("\nAIS Model Training Metrics:")

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R2 Score:", r2)

# Individual target metrics

allocated_mae = mean_absolute_error(
    y[:, 0],
    ais_predictions[:, 0]
)

utilized_mae = mean_absolute_error(
    y[:, 1],
    ais_predictions[:, 1]
)

allocated_r2 = r2_score(
    y[:, 0],
    ais_predictions[:, 0]
)

utilized_r2 = r2_score(
    y[:, 1],
    ais_predictions[:, 1]
)

# ============================================================
# 14. AIS ACCURACY GRAPH
# ============================================================

# Regression uses R2 rather than classification accuracy.

metric_names = [
    "Allocated Funds",
    "Utilized Funds",
    "Overall"
]

metric_values = [
    allocated_r2,
    utilized_r2,
    r2
]

plt.figure(
    figsize=(10, 6)
)

bars = plt.bar(
    metric_names,
    metric_values,
    color=[
        "royalblue",
        "darkorange",
        "seagreen"
    ]
)

plt.axhline(
    0,
    color="black",
    linewidth=0.8
)

plt.title(
    "AIS Model Training Fit (R²)",
    fontsize=14
)

plt.xlabel("Prediction Targets")
plt.ylabel("R² Score")

plt.grid(
    axis="y",
    alpha=0.3
)

for bar, value in zip(
    bars,
    metric_values
):

    plt.text(
        bar.get_x()
        + bar.get_width() / 2,
        value,
        f"{value:.4f}",
        ha="center",
        va="bottom"
        if value >= 0
        else "top"
    )

plt.tight_layout()

plt.savefig(
    output_path(
        "accuracy_graph.png"
    ),
    dpi=300
)

plt.show()

# ============================================================
# 15. AIS COMPARISON GRAPH
# ============================================================

comparison_df = pd.DataFrame({

    "Metric": [
        "MAE",
        "RMSE"
    ],

    "Value": [
        mae,
        rmse
    ]

})

plt.figure(
    figsize=(9, 6)
)

sns.barplot(
    data=comparison_df,
    x="Metric",
    y="Value",
    color="steelblue"
)

plt.title(
    "AIS Model Error Comparison"
)

plt.xlabel("Evaluation Metric")
plt.ylabel("Error Value")

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    output_path(
        "comparison_graph.png"
    ),
    dpi=300
)

plt.show()

# ============================================================
# 16. CORRELATION HEATMAP
# ============================================================

correlation_df = train_df[
    [
        "Year_Numeric",
        allocated_col,
        utilized_col
    ]
]

correlation_matrix = correlation_df.corr()

plt.figure(
    figsize=(8, 6)
)

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title(
    "AIS Government Fund Correlation Heatmap"
)

plt.tight_layout()

plt.savefig(
    output_path(
        "heatmap.png"
    ),
    dpi=300
)

plt.show()

# ============================================================
# 17. HISTORICAL RESULTS CSV
# ============================================================

result_df = pd.DataFrame({

    "Year": years.astype(int),

    "Actual_Allocated":
        y[:, 0],

    "Predicted_Allocated":
        ais_predictions[:, 0],

    "Actual_Utilized":
        y[:, 1],

    "Predicted_Utilized":
        ais_predictions[:, 1]

})

result_df[
    "Allocated_Error"
] = (
    result_df["Actual_Allocated"]
    - result_df["Predicted_Allocated"]
)

result_df[
    "Utilized_Error"
] = (
    result_df["Actual_Utilized"]
    - result_df["Predicted_Utilized"]
)

result_df.to_csv(
    output_path(
        "result.csv"
    ),
    index=False
)

print("\nAIS Historical Results:")
print(result_df)

# ============================================================
# 18. HISTORICAL RESULT GRAPH
# ============================================================

plt.figure(
    figsize=(12, 6)
)

plt.plot(
    result_df["Year"],
    result_df["Actual_Allocated"],
    marker="o",
    linewidth=2,
    label="Actual Allocated"
)

plt.plot(
    result_df["Year"],
    result_df["Predicted_Allocated"],
    marker="o",
    linestyle="--",
    label="AIS Predicted Allocated"
)

plt.plot(
    result_df["Year"],
    result_df["Actual_Utilized"],
    marker="s",
    linewidth=2,
    label="Actual Utilized"
)

plt.plot(
    result_df["Year"],
    result_df["Predicted_Utilized"],
    marker="s",
    linestyle="--",
    label="AIS Predicted Utilized"
)

plt.title(
    "AIS Actual vs Predicted Government Funds"
)

plt.xlabel("Year")
plt.ylabel("Funds")

plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()

plt.savefig(
    output_path(
        "result_graph.png"
    ),
    dpi=300
)

plt.show()

# ============================================================
# 19. FUTURE FUND FORECASTING
# ============================================================

last_year = int(
    years.max()
)

future_years = np.arange(
    last_year + 1,
    last_year + 6
).reshape(-1, 1)

future_scaled = x_scaler.transform(
    future_years
)

future_predictions_scaled = ais_predict_scaled(
    best_antibody,
    future_scaled
)

future_predictions = y_scaler.inverse_transform(
    future_predictions_scaled
)

# ============================================================
# 20. FUTURE PREDICTION CSV
# ============================================================

prediction_df = pd.DataFrame({

    "Year":
        future_years.flatten(),

    "Predicted_Allocated":
        future_predictions[:, 0],

    "Predicted_Utilized":
        future_predictions[:, 1]

})

prediction_df[
    "Financial_Year"
] = (
    prediction_df["Year"]
    .astype(int)
    .astype(str)
    + "-"
    + (
        prediction_df["Year"].astype(int) + 1
    ).astype(str).str[-2:]
)

prediction_df[
    "Predicted_Utilization_Percentage"
] = np.where(
    prediction_df["Predicted_Allocated"] != 0,
    (
        prediction_df["Predicted_Utilized"]
        / prediction_df["Predicted_Allocated"]
    ) * 100,
    np.nan
)

prediction_df.to_csv(
    output_path(
        "prediction.csv"
    ),
    index=False
)

print("\nAIS Future Predictions:")
print(prediction_df)

# ============================================================
# 21. FUTURE PREDICTION GRAPH
# ============================================================

plt.figure(
    figsize=(12, 6)
)

plt.plot(
    years,
    y[:, 0],
    marker="o",
    linewidth=2,
    label="Historical Allocated"
)

plt.plot(
    years,
    y[:, 1],
    marker="s",
    linewidth=2,
    label="Historical Utilized"
)

plt.plot(
    prediction_df["Year"],
    prediction_df["Predicted_Allocated"],
    marker="o",
    linestyle="--",
    linewidth=2,
    label="AIS Forecast Allocated"
)

plt.plot(
    prediction_df["Year"],
    prediction_df["Predicted_Utilized"],
    marker="s",
    linestyle="--",
    linewidth=2,
    label="AIS Forecast Utilized"
)

plt.title(
    "AIS Government Fund Forecasting"
)

plt.xlabel("Year")
plt.ylabel("Funds")

plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()

plt.savefig(
    output_path(
        "prediction_graph.png"
    ),
    dpi=300
)

plt.show()

# ============================================================
# 22. AIS CONVERGENCE GRAPH
# ============================================================

optimization_df = pd.DataFrame({

    "Generation":
        np.arange(
            1,
            GENERATIONS + 1
        ),

    "Best_Fitness":
        fitness_history,

    "Average_Fitness":
        average_fitness_history

})

optimization_df.to_csv(
    output_path(
        "optimization_history.csv"
    ),
    index=False
)

plt.figure(
    figsize=(12, 6)
)

plt.plot(
    optimization_df["Generation"],
    optimization_df["Best_Fitness"],
    linewidth=2,
    label="Best Fitness"
)

plt.plot(
    optimization_df["Generation"],
    optimization_df["Average_Fitness"],
    linestyle="--",
    label="Average Fitness"
)

plt.title(
    "AIS Clonal Selection Optimization Convergence"
)

plt.xlabel("Generation")
plt.ylabel("Fitness (Loss)")

plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()

plt.savefig(
    output_path(
        "convergence_graph.png"
    ),
    dpi=300
)

plt.show()

# ============================================================
# 23. FUND UTILIZATION ANALYSIS
# ============================================================

utilization_df = pd.DataFrame({

    "Year":
        years.astype(int),

    "Funds_Allocated":
        y[:, 0],

    "Funds_Utilized":
        y[:, 1]

})

utilization_df[
    "Utilization_Percentage"
] = np.where(
    utilization_df["Funds_Allocated"] != 0,
    (
        utilization_df["Funds_Utilized"]
        / utilization_df["Funds_Allocated"]
    ) * 100,
    np.nan
)

utilization_df[
    "Fund_Gap"
] = (
    utilization_df["Funds_Allocated"]
    - utilization_df["Funds_Utilized"]
)

utilization_df.to_csv(
    output_path(
        "utilization_analysis.csv"
    ),
    index=False
)

plt.figure(
    figsize=(10, 6)
)

sns.barplot(
    data=utilization_df,
    x="Year",
    y="Utilization_Percentage",
    color="teal"
)

plt.axhline(
    100,
    linestyle="--",
    color="red",
    label="100% Utilization"
)

plt.title(
    "Government Fund Utilization Efficiency"
)

plt.xlabel("Year")
plt.ylabel("Utilization Percentage (%)")

plt.legend()
plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    output_path(
        "utilization_graph.png"
    ),
    dpi=300
)

plt.show()

# ============================================================
# 24. SAVE AIS MODEL AS PKL
# ============================================================

ais_model_data = {

    "algorithm":
        "Artificial Immune System - Clonal Selection",

    "best_antibody":
        best_antibody,

    "polynomial_degree":
        POLYNOMIAL_DEGREE,

    "n_features":
        N_FEATURES,

    "n_targets":
        N_TARGETS,

    "x_scaler":
        x_scaler,

    "y_scaler":
        y_scaler,

    "target_columns":
        [allocated_col, utilized_col]

}

joblib.dump(
    ais_model_data,
    output_path(
        "model.pkl"
    )
)

print("\nAIS PKL Model Saved!")

# ============================================================
# 25. SAVE AIS MODEL AS H5
# ============================================================

# This H5 file stores AIS-optimized regression weights.
# It is not a Keras neural network model.

with h5py.File(
    output_path(
        "model.h5"
    ),
    "w"
) as h5_file:

    h5_file.create_dataset(
        "best_antibody",
        data=best_antibody
    )

    h5_file.create_dataset(
        "weights",
        data=decode_antibody(
            best_antibody
        )
    )

    h5_file.create_dataset(
        "x_mean",
        data=x_scaler.mean_
    )

    h5_file.create_dataset(
        "x_scale",
        data=x_scaler.scale_
    )

    h5_file.create_dataset(
        "y_mean",
        data=y_scaler.mean_
    )

    h5_file.create_dataset(
        "y_scale",
        data=y_scaler.scale_
    )

    h5_file.attrs[
        "algorithm"
    ] = "AIS Clonal Selection"

    h5_file.attrs[
        "polynomial_degree"
    ] = POLYNOMIAL_DEGREE

print("\nAIS H5 Model Saved!")

# ============================================================
# 26. SAVE YAML CONFIGURATION
# ============================================================

config = {

    "project":
        "AI-Based Government Fund Allocation and Utilization Analysis and Forecasting",

    "algorithm":
        "Artificial Immune System",

    "optimization_method":
        "Clonal Selection",

    "dataset":
        os.path.basename(
            CSV_PATH
        ),

    "total_records":
        int(len(df)),

    "training_records":
        int(len(train_df)),

    "features":
        ["Year_Numeric"],

    "targets":
        [allocated_col, utilized_col],

    "model": {

        "type":
            "Polynomial Regression",

        "degree":
            POLYNOMIAL_DEGREE,

        "parameters":
            N_PARAMETERS

    },

    "ais_parameters": {

        "population_size":
            POPULATION_SIZE,

        "generations":
            GENERATIONS,

        "selection_size":
            SELECTION_SIZE,

        "clone_factor":
            CLONE_FACTOR,

        "mutation_rate":
            MUTATION_RATE,

        "random_replacement":
            RANDOM_REPLACEMENT

    },

    "evaluation":
        "Training-fit metrics only",

    "forecast_reliability":
        "Experimental - insufficient historical data"

}

with open(
    output_path(
        "config.yaml"
    ),
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        config,
        file,
        sort_keys=False
    )

print("\nAIS YAML Configuration Saved!")

# ============================================================
# 27. SAVE JSON RESULTS
# ============================================================

json_results = {

    "project":
        "Government Fund Allocation and Utilization Forecasting",

    "algorithm":
        "Artificial Immune System",

    "method":
        "Clonal Selection",

    "dataset_summary": {

        "total_records":
            int(len(df)),

        "training_records":
            int(len(train_df))

    },

    "model_metrics": {

        "MAE":
            float(mae),

        "MSE":
            float(mse),

        "RMSE":
            float(rmse),

        "R2":
            float(r2),

        "Allocated_MAE":
            float(allocated_mae),

        "Utilized_MAE":
            float(utilized_mae),

        "Allocated_R2":
            float(allocated_r2),

        "Utilized_R2":
            float(utilized_r2)

    },

    "optimization": {

        "best_fitness":
            float(best_fitness),

        "generations":
            GENERATIONS,

        "population_size":
            POPULATION_SIZE

    },

    "evaluation_type":
        "In-sample training evaluation",

    "future_predictions":
        prediction_df.replace(
            [np.inf, -np.inf],
            np.nan
        ).where(
            pd.notnull(prediction_df),
            None
        ).to_dict(
            orient="records"
        )

}

with open(
    output_path(
        "results.json"
    ),
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        json_results,
        file,
        indent=4
    )

print("\nAIS JSON Results Saved!")

# ============================================================
# 28. FINAL OUTPUT SUMMARY
# ============================================================

print("\n" + "=" * 65)

print(
    "ARTIFICIAL IMMUNE SYSTEM ANALYSIS COMPLETED"
)

print("=" * 65)

generated_files = [

    "accuracy_graph.png",
    "comparison_graph.png",
    "heatmap.png",
    "result.csv",
    "result_graph.png",
    "prediction.csv",
    "prediction_graph.png",
    "convergence_graph.png",
    "utilization_graph.png",
    "utilization_analysis.csv",
    "optimization_history.csv",
    "model.pkl",
    "model.h5",
    "scalers.pkl",
    "config.yaml",
    "results.json"

]

print("\nGenerated Files:\n")

for filename in generated_files:

    full_path = output_path(
        filename
    )

    if os.path.exists(
        full_path
    ):

        print(
            "[SAVED] ais_" + filename
        )

    else:

        print(
            "[MISSING] ais_" + filename
        )

print("\nOutput Directory:")
print(PROJECT_DIR)

print("\nAIS Training Metrics:")

print(f"MAE  : {mae:.6f}")
print(f"MSE  : {mse:.6f}")
print(f"RMSE : {rmse:.6f}")
print(f"R2   : {r2:.6f}")

print("\nBest AIS Fitness:")
print(best_fitness)

print("\nIMPORTANT:")
print(
    "These are training-fit metrics, "
    "not independently validated accuracy."
)

print(
    "Future forecasts are experimental "
    "because the dataset is very small."
)

print("\nPROGRAM EXECUTION COMPLETED SUCCESSFULLY!")
```
